3. Create a numpy array of ones and a numpy array of zeroes

```python
ones = np.ones((5, 5))
zeros = np.zeros((3, 6))

print("Numpy array of ones:")
print(ones)

print("\nNumpy array of zeros:")
print(zeros)
```

4. Create a random numpy array of size 6x3 and transpose it

```python
rand_arr = np.random.rand(6, 3)
rand_arr_trans = rand_arr.T

print("Random numpy array of size 6x3:")
print(rand_arr)
print("\nRandom numpy array of size 6x3 after transposed:")
print(rand_arr_trans)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n = 2
p = 3
m = 4

array_A = np.random.rand(n, p)
array_B = np.random.rand(p, m)

print("Array A:")
print(array_A)

print("Array B:")
print(array_B)

dot_product = np.dot(array_A, array_B)

print("Dot product:")
print(dot_product)
```

or

```python
n = 2
p = 3
m = 4

array_A = np.random.rand(n, p)
array_B = np.random.rand(p, m)

print("Array A:")
print(array_A)

print("Array B:")
print(array_B)

print("Dot product:")
print(array_A @ array_B)
print("\nResult shape: ", dot_product.shape)
```
